# L-Systems (Lindenmayer Systems) Lab

## Introduction

L-Systems are parallel rewriting systems that were introduced by Aristid Lindenmayer in 1968. They are particularly useful for modeling plant growth and generating fractals. In this lab, we'll explore how to create various patterns using L-Systems.

### Key Concepts:
- **Axiom**: The initial state/string
- **Production Rules**: Rules that define how to replace characters
- **Iterations**: Number of times to apply the rules
- **Turtle Graphics**: System for visualizing the L-System output

## Part 1: Basic L-System Implementation

First, let's implement our core L-System functions. These will be used throughout the lab to generate and draw various patterns.

```python
def create_l_system(iterations, axiom, rules):
    """Generate L-System instructions based on axiom and rules."""
    result = axiom
    for _ in range(iterations):
        new_string = ""
        for char in result:
            new_string += rules.get(char, char)
        result = new_string
    return result

def draw_l_system(instructions, angle, distance):
    """Draw the L-System using turtle graphics.

    Parameters:
    - instructions: string of L-System commands
    - angle: turning angle in degrees
    - distance: forward movement distance
    """
    stack = []
    for cmd in instructions:
        if cmd == 'F':  # Move forward and draw
            t.forward(distance)
        elif cmd == 'f':  # Move forward without drawing
            t.penup()
            t.forward(distance)
            t.pendown()
        elif cmd == '+':  # Turn right
            t.right(angle)
        elif cmd == '-':  # Turn left
            t.left(angle)
        elif cmd == '[':  # Save current state
            stack.append((t.position(), t.heading()))
        elif cmd == ']':  # Restore previous state
            position, heading = stack.pop()
            t.penup()
            t.goto(position)
            t.setheading(heading)
            t.pendown()

def setup_turtle():
        t.initializeTurtle()
        t.hideturtle()
        t.speed(13)  # Fastest speed
        t.penup()
        t.goto(t.window_width() // 2, t.window_height() - 50)  # Start position
        t.pendown()
```

## Exercise 1: Koch Curve

The Koch curve is a classic example of a fractal pattern. Let's create it using our L-System:

```python
# Koch curve parameters
koch_axiom = "F"
koch_rules = {"F": "F+F-F-F+F"}
koch_iterations = 3
koch_angle = 90

# Generate and draw
setup_turtle()
t.pensize(2)
koch_instructions = create_l_system(koch_iterations, koch_axiom, koch_rules)
draw_l_system(koch_instructions, koch_angle, 5)
```

Assignment Project: Between the Abyss and the Stars

This project uses L-Systems to explore visual similarities between
organic structures in the deep ocean and large-scale structures in space.

Two different L-System rule sets are used:
1. Bioluminescent Coral
2. Cosmic Bloom

The generated structures can be changed through parameters such as
iteration depth, turning angle, segment length, color, and line thickness.

```python
def draw_growth_system(instructions, angle, distance,
                       colors, thicknesses):
    """Draw an L-System with colors and line thickness
    based on branch depth."""

    stack = []
    depth = 0

    for cmd in instructions:

        # Choose appearance based on current branch depth
        level = min(depth, len(colors) - 1)
        t.color(colors[level])
        t.pensize(thicknesses[level])

        if cmd == 'F':
            t.forward(distance)

        elif cmd == 'f':
            t.penup()
            t.forward(distance)
            t.pendown()

        elif cmd == '+':
            t.right(angle)

        elif cmd == '-':
            t.left(angle)

        elif cmd == '[':
            stack.append((t.position(), t.heading(), depth))
            depth += 1

        elif cmd == ']':
            position, heading, depth = stack.pop()
            t.penup()
            t.goto(position)
            t.setheading(heading)
            t.pendown()
```

```python
# Pattern 1: Bioluminescent Coral

coral_axiom = "F"

coral_rules = {
    "F": "F[+F][--F]F[+F]"
}

coral_iterations = 4
coral_angle = 28
coral_distance = 9

coral_colors = [
    "#183D5D",
    "#246B8E",
    "#32B7C7",
    "#82E6E6",
    "#E5FFFF"
]

coral_thickness = [6, 4, 3, 2, 1]
```

```python
setup_turtle()

coral_instructions = create_l_system(
    coral_iterations,
    coral_axiom,
    coral_rules
)

draw_growth_system(
    coral_instructions,
    coral_angle,
    coral_distance,
    coral_colors,
    coral_thickness
)
```

## Pattern 2: Stellar Bloom

This pattern uses a radial L-System inspired by snowflakes, crystals, and stars. The structure grows outward from the center, creating a more symmetrical pattern than the coral.

```python
## Pattern 2: Stellar Bloom

bloom_axiom = "[X][+X][++X][+++X][++++X][+++++X][++++++X]"

bloom_rules = {
    "X": "F[+X][-X]FX",
    "F": "FF"
}

bloom_iterations = 4
bloom_angle = 51.4
bloom_distance = 4

bloom_colors = [
    "#4B3F72",
    "#6D5BA8",
    "#9B7EDE",
    "#C9B6FF",
    "#F4EEFF"
]

bloom_thickness = [4, 3, 2, 1, 1]
```

```python
def setup_cosmic_turtle():
    t.initializeTurtle(initial_speed=10)

    t.bgcolor("black")

    t.penup()
    t.goto(
        t.window_width() // 2,
        t.window_height() // 2
    )

    t.setheading(0)
    t.pendown()
    t.hideturtle()
```

```python
setup_cosmic_turtle()

bloom_instructions = create_l_system(
    bloom_iterations,
    bloom_axiom,
    bloom_rules
)

draw_growth_system(
    bloom_instructions,
    bloom_angle,
    bloom_distance,
    bloom_colors,
    bloom_thickness
)
```

## Pattern 3: Cosmic Garden

This pattern combines both L-System rule sets in one scene. Multiple coral-like structures grow from different positions, while a Stellar Bloom appears above them. Small points of light are added to make the environment feel somewhere between the deep ocean and outer space.

```python
import random

def draw_stars(count=60):
    """Draw small random points of light in the background."""

    star_colors = ["#FFFFFF", "#C9E8FF", "#D8CCFF"]

    for _ in range(count):
        x = random.randint(30, t.window_width() - 30)
        y = random.randint(30, t.window_height() - 30)

        t.penup()
        t.goto(x, y)
        t.pendown()

        t.color(random.choice(star_colors))
        t.pensize(random.choice([1, 1, 1, 2]))

        # very small line used as a point of light
        t.forward(random.choice([1, 2]))
```

```python
def move_turtle(x, y, heading):
    t.penup()
    t.goto(x, y)
    t.setheading(heading)
    t.pendown()
```

### Sample 5: Cosmic Garden

This final example combines both L-System rule sets in the same environment. Two coral structures use different iteration depths and angles, while a Stellar Bloom appears in the distance. Random points of light make the scene ambiguous between the deep ocean and outer space.

```python
# Pattern 3: Cosmic Garden

t.initializeTurtle(initial_speed=10)
t.bgcolor("black")
t.hideturtle()

# Background
draw_stars(70)

# Create the L-System instructions
coral_instructions = create_l_system(
    coral_iterations,
    coral_axiom,
    coral_rules
)

bloom_instructions = create_l_system(
    bloom_iterations,
    bloom_axiom,
    bloom_rules
)

# Coral 1
coral_large = create_l_system(
    4,
    coral_axiom,
    coral_rules
)

move_turtle(245, 475, 270)

draw_growth_system(
    coral_large,
    25,
    7,
    coral_colors,
    coral_thickness
)

# Coral 2
coral_small = create_l_system(
    3,
    coral_axiom,
    coral_rules
)

move_turtle(535, 480, 265)

draw_growth_system(
    coral_small,
    31,
    8,
    coral_colors,
    coral_thickness
)

# Stellar Bloom
move_turtle(385, 165, 0)

draw_growth_system(
    bloom_instructions,
    bloom_angle,
    2.7,
    bloom_colors,
    bloom_thickness
)
```

# Technical Documentation

This project uses two different L-System rule sets to generate visual structures inspired by natural and cosmic forms. The first system, Bioluminescent Coral, grows upward from a single starting point using the rule `F → F[+F][--F]F[+F]`. The second system, Stellar Bloom, uses both `X` and `F` symbols to create a radial structure that expands outward from the center.

The L-System strings are generated by repeatedly applying production rules for a selected number of iterations. The resulting strings are then interpreted using turtle graphics. `F` moves the turtle forward, `+` and `-` change its direction, and `[` and `]` save and restore its position and heading.

I modified the drawing process to track branch depth. This allows color and line thickness to change without changing the underlying L-System rules. Older parts of the structure are darker and thicker, while newer branches become lighter and thinner.

The main adjustable parameters are iteration depth, turning angle, and segment length. Changing these values can significantly affect the density, size, and shape of the generated patterns. One challenge was finding rules that produced complex structures without becoming too dense. I adjusted the production rules and drawing parameters through several experiments until the coral and radial patterns remained visually distinct.
